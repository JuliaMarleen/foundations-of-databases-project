```python
import pandas as pd
from sqlalchemy import create_engine

# Read the CSV file into a DataFrame
df_airports = pd.read_csv('airports.csv')
df_countries = pd.read_csv('countries.csv')
df_zones = pd.read_csv('zones.csv')

# Create new database
engine = create_engine('sqlite:///assignment_airport.db')

# Import data into new table in SQLite database
df_airports.to_sql('airports', con=engine, if_exists='replace', index=False)
df_countries.to_sql('countries', con=engine, if_exists='replace', index=False)
df_zones.to_sql('zones', con=engine, if_exists='replace', index=False)
```

```python
%%sql
/* Question 4A */
/* Select all airports with latitude higher than 50 and only show the first (id order) airport per country */
/* For if you want to select a flight to a place that is more northern than the 50 degree latitude, but you do 
not want to see every airport */

SELECT city, name, latitude, country_id
FROM airports
GROUP BY country_id
HAVING latitude > 50
ORDER BY latitude;
```

```python
%%sql
/* This is the list of countries you can get with the last query, so you know which countries you can visit with
a latitude higher than 50 degrees.*/
SELECT name, continent
FROM countries
WHERE id
IN (SELECT country_id
FROM airports
GROUP BY country_id
HAVING latitude > 50);
```

```python
%%sql
/* Question 4B */
/* Select all airports that are not in Asia or Europe, for if you want to fly anywhere but there. */
SELECT name, city, country_id
FROM airports
WHERE country_id
NOT IN (SELECT id FROM countries WHERE continent = 'Asia' OR continent = 'Europe')
ORDER BY country_id;
```

```python
%%sql
/* Question 4C */
/* Devide all countries by currency, count how much countries have that currency and order them by name. */
SELECT name, currency_code, continent,
       COUNT(currency_code) OVER (PARTITION BY currency_code ORDER BY name) AS cumulative_countries_currency
FROM countries;
```

```python
%%sql
/* Not a Question, but an otherwise unused table */
/* Select all airports that are in zone 1 and because of that have 20 percent discount */
/* For if you want to travel cheap */
SELECT name, continent, zone_id
FROM countries
WHERE zone_id
IN (SELECT id FROM zones WHERE discount_percentage = 20)
ORDER BY name;
```
